```python
import fitz  # PyMuPDF
import io
import os
from PIL import Image
import tempfile

def extract_images_from_pdf(pdf_path, output_folder):
    # Create output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # Open the PDF
    pdf_document = fitz.open(pdf_path)
    
    # Counter for extracted images
    image_count = 0
    
    # Iterate through each page
    for page_num in range(len(pdf_document)):
        page = pdf_document[page_num]
        
        # Method 1: Try to extract images directly
        image_list = page.get_images(full=True)
        
        for img_index, img in enumerate(image_list):
            try:
                xref = img[0]  # Get the XREF of the image
                base_image = pdf_document.extract_image(xref)
                image_bytes = base_image["image"]
                
                # Try to convert to JPG using PIL
                try:
                    # Convert to JPG using PIL
                    image = Image.open(io.BytesIO(image_bytes))
                    image_filename = f"{os.path.basename(pdf_path)}_page{page_num+1}_img{img_index+1}.jpg"
                    output_path = os.path.join(output_folder, image_filename)
                    
                    # Convert to RGB if the image is in RGBA mode (JPG doesn't support alpha channel)
                    if image.mode == 'RGBA':
                        image = image.convert('RGB')
                    
                    image.save(output_path, "JPEG", quality=95)
                    image_count += 1
                    print(f"Extracted and converted image {img_index+1} from page {page_num+1} to JPG")
                    
                except Exception as e:
                    print(f"Error converting image {img_index+1} on page {page_num+1} to JPG: {str(e)}")
                    # If PIL conversion fails, try to save in original format
                    image_ext = base_image["ext"]
                    image_filename = f"{os.path.basename(pdf_path)}_page{page_num+1}_img{img_index+1}.{image_ext}"
                    output_path = os.path.join(output_folder, image_filename)
                    
                    with open(output_path, "wb") as img_file:
                        img_file.write(image_bytes)
                    
                    print(f"Saved image {img_index+1} from page {page_num+1} in original format ({image_ext})")
                    image_count += 1
                
            except Exception as e:
                print(f"Error extracting image {img_index+1} on page {page_num+1}: {str(e)}")
                
        # Method 2: Alternative approach - render page as image
        try:
            # Higher zoom factor for better resolution (2.0 = 2x resolution)
            zoom_factor = 2.0
            mat = fitz.Matrix(zoom_factor, zoom_factor)
            pix = page.get_pixmap(matrix=mat)
            
            # Save the rendered page as JPG
            page_image_filename = f"{os.path.basename(pdf_path)}_page{page_num+1}_full.jpg"
            page_output_path = os.path.join(output_folder, page_image_filename)
            
            # Convert pixmap to PIL Image and save as JPG
            img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
            img.save(page_output_path, "JPEG", quality=95)
            
            print(f"Rendered full page {page_num+1} as JPG image")
            
        except Exception as e:
            print(f"Error rendering page {page_num+1}: {str(e)}")
    
    print(f"Extracted {image_count} individual images and rendered {len(pdf_document)} pages from {pdf_path}")
    pdf_document.close()

# Example usage
pdf_file = r"D:\Gen AI\Video Class\C06 Triangles.pdf"
output_dir = r"D:\Gen AI\Video Class\Gr10 Images"
extract_images_from_pdf(pdf_file, output_dir)
```

```python
import fitz  # PyMuPDF
import io
import os
from PIL import Image
import numpy as np

def extract_images_from_pdf(pdf_path, output_folder):
    # Create output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # Open the PDF
    pdf_document = fitz.open(pdf_path)
    
    # Counter for extracted images
    image_count = 0
    
    # Iterate through each page
    for page_num in range(len(pdf_document)):
        page = pdf_document[page_num]
        
        # APPROACH 1: Extract images using PyMuPDF's built-in methods
        image_list = page.get_images(full=True)
        
        for img_index, img in enumerate(image_list):
            try:
                xref = img[0]  # Get the XREF of the image
                base_image = pdf_document.extract_image(xref)
                image_bytes = base_image["image"]
                
                # Try to convert to JPG using PIL
                try:
                    # Open image with PIL
                    image = Image.open(io.BytesIO(image_bytes))
                    
                    # Check if image is mostly black (potential issue)
                    is_mostly_black = False
                    if image.mode in ['RGB', 'RGBA', 'L']:
                        img_array = np.array(image)
                        black_pixels = np.sum(img_array < 30)  # Count dark pixels
                        total_pixels = img_array.size
                        if black_pixels / total_pixels > 0.95:  # If more than 95% is black
                            is_mostly_black = True
                            print(f"Warning: Image {img_index+1} on page {page_num+1} appears to be mostly black")
                    
                    # Convert to RGB if needed
                    if image.mode == 'RGBA':
                        # Use white background for transparency
                        background = Image.new('RGB', image.size, (255, 255, 255))
                        background.paste(image, mask=image.split()[3])  # 3 is the alpha channel
                        image = background
                    elif image.mode != 'RGB':
                        image = image.convert('RGB')
                    
                    # Skip saving if image is too small or mostly black
                    if image.width < 10 or image.height < 10 or is_mostly_black:
                        print(f"Skipping low-quality image {img_index+1} on page {page_num+1}")
                        continue
                    
                    image_filename = f"{os.path.basename(pdf_path)}_page{page_num+1}_img{img_index+1}.jpg"
                    output_path = os.path.join(output_folder, image_filename)
                    image.save(output_path, "JPEG", quality=95)
                    image_count += 1
                    print(f"Extracted image {img_index+1} from page {page_num+1}")
                    
                except Exception as e:
                    print(f"Error processing image {img_index+1} on page {page_num+1}: {str(e)}")
            
            except Exception as e:
                print(f"Error extracting image {img_index+1} on page {page_num+1}: {str(e)}")
        
        # APPROACH 2: Render page regions as images
        # This approach is better for diagrams, charts, and other content
        try:
            # Higher zoom factor for better resolution
            zoom_factor = 3.0  # Increased for better quality
            mat = fitz.Matrix(zoom_factor, zoom_factor)
            
            # Get the page's rect object
            rect = page.rect
            
            # Render the page
            pix = page.get_pixmap(matrix=mat, alpha=False)  # alpha=False for JPG compatibility
            
            # Convert to PIL Image
            img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
            
            # Save the full page
            page_image_filename = f"{os.path.basename(pdf_path)}_page{page_num+1}_full.jpg"
            page_output_path = os.path.join(output_folder, page_image_filename)
            img.save(page_output_path, "JPEG", quality=95)
            
            print(f"Rendered full page {page_num+1} as image")
            
            # APPROACH 3: Try to detect and extract figures/diagrams from the page
            # This uses PyMuPDF's ability to find image-like regions
            blocks = page.get_text("dict")["blocks"]
            for i, block in enumerate(blocks):
                # Check if block is an image
                if block["type"] == 1:  # Type 1 is image
                    try:
                        # Get the image rectangle and adjust by zoom factor
                        img_rect = fitz.Rect(block["bbox"])
                        # Add some padding around the image
                        img_rect.x0 -= 5
                        img_rect.y0 -= 5
                        img_rect.x1 += 5
                        img_rect.y1 += 5
                        
                        # Make sure rect is within page bounds
                        img_rect = img_rect.intersect(rect)
                        
                        # Skip if too small
                        if img_rect.width < 20 or img_rect.height < 20:
                            continue
                            
                        # Render just this part of the page
                        clip_pix = page.get_pixmap(matrix=mat, clip=img_rect, alpha=False)
                        
                        # Convert to PIL Image
                        clip_img = Image.frombytes("RGB", [clip_pix.width, clip_pix.height], clip_pix.samples)
                        
                        # Check if image is mostly white or black (might be a false positive)
                        img_array = np.array(clip_img)
                        white_pixels = np.sum(img_array > 240)
                        black_pixels = np.sum(img_array < 30)
                        total_pixels = img_array.size
                        
                        # Skip if the image is mostly white or black
                        if white_pixels / total_pixels > 0.95 or black_pixels / total_pixels > 0.95:
                            continue
                        
                        # Save the image
                        region_filename = f"{os.path.basename(pdf_path)}_page{page_num+1}_region{i}.jpg"
                        region_path = os.path.join(output_folder, region_filename)
                        clip_img.save(region_path, "JPEG", quality=95)
                        print(f"Extracted region {i} from page {page_num+1}")
                        
                    except Exception as e:
                        print(f"Error extracting region {i} from page {page_num+1}: {str(e)}")
            
        except Exception as e:
            print(f"Error processing page {page_num+1}: {str(e)}")
    
    print(f"Extraction complete. Processed {len(pdf_document)} pages from {pdf_path}")
    pdf_document.close()

# Example usage
pdf_file = r"D:\Gen AI\Video Class\C06 Triangles.pdf"
output_dir = r"D:\Gen AI\Video Class\Gr10 Images"
extract_images_from_pdf(pdf_file, output_dir)
```

```python
from pdf2docx import Converter
import os

def convert_pdf_to_word(pdf_path, output_folder=None):
    """
    Convert a PDF file to a Word document
    
    Args:
        pdf_path (str): Path to the PDF file
        output_folder (str, optional): Folder to save the Word document. 
                                      If None, saves in the same folder as the PDF.
    
    Returns:
        str: Path to the created Word document
    """
    # Get the base filename without extension
    base_name = os.path.basename(pdf_path)
    file_name = os.path.splitext(base_name)[0]
    
    # Determine output folder
    if output_folder is None:
        output_folder = os.path.dirname(pdf_path)
    
    # Create output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # Create output file path
    docx_path = os.path.join(output_folder, f"{file_name}.docx")
    
    # Convert PDF to Word
    try:
        # Initialize the converter
        cv = Converter(pdf_path)
        
        # Convert to Word
        print(f"Converting {pdf_path} to Word...")
        cv.convert(docx_path, start=0, end=None)
        
        # Close the converter
        cv.close()
        
        print(f"Conversion complete! Word document saved at: {docx_path}")
        return docx_path
    
    except Exception as e:
        print(f"Error converting PDF to Word: {str(e)}")
        return None

# Example usage
pdf_file = r"D:\Gen AI\Video Class\C06 Triangles.pdf"
output_dir = r"D:\Gen AI\Video Class\Gr10 Word"

# Install required package if not already installed
try:
    import pdf2docx
except ImportError:
    print("Installing required package: pdf2docx")
    import pip
    pip.main(['install', 'pdf2docx'])
    print("Package installed successfully!")

# Convert PDF to Word
convert_pdf_to_word(pdf_file, output_dir)
```

```python
import PyPDF2
from nltk.tokenize import word_tokenize

# Define the list of files
files = [r"D:\Gen AI\GenAI Prac\Source\nltk files\AnnualHealthCheck.pdf",
         r"D:\Gen AI\GenAI Prac\Source\nltk files\LeavePolicy.pdf",
         r"D:\Gen AI\GenAI Prac\Source\nltk files\NoticePeriod.pdf",
         r"D:\Gen AI\GenAI Prac\Source\nltk files\OfficeTime.pdf",
         r"D:\Gen AI\GenAI Prac\Source\nltk files\Separation.pdf",
         r"D:\Gen AI\GenAI Prac\Source\nltk files\Travel.pdf",
         r"D:\Gen AI\GenAI Prac\Source\nltk files\USA_Employee_Handbook-Freely_Available.pdf"]

x = []  # This will store the extracted text from PDFs

# Process each PDF file
for file in files:
    reader = PyPDF2.PdfReader(file)
    text = ""
    for page in reader.pages:
        text += page.extract_text()
    x.append(text)

print("Documents Loaded:", len(x))

# Get user query
query = input("Ask Question: ")

# Tokenize the query
query_tokens = word_tokenize(query)
print("Query Tokens:", query_tokens)

# Compare query tokens with document tokens
for doc in x:
    tokens = word_tokenize(doc)
    
    common = set(tokens).intersection(set(query_tokens))
    print("Common Tokens:", common)

    if len(common) > 0:
        print("\nRelevant Section:\n")
        print(doc[:500])  # Print first 500 characters of matching document

# To delete old data stored, simply reassign the variable
x = []  # This clears the list of document texts
print("Data cleared. Length of x:", len(x))
```
