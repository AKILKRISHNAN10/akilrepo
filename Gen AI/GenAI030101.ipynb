```python
import re

def format_number(num):
    # Remove non-digits
    num = re.sub(r"\D", "",num)
    
    

    # Handle different cases
    if num.startswith("91") and len(num) == 12:
        return "+91" + num[2:]
    elif len(num) == 10:
        return "+91" + num
    else:
        return "Invalid"


for n in t:
    print(format_number(n))

valid_numbers = []

for n in t:
    cleaned = format_number(n)

    if cleaned and re.fullmatch(valid_form,cleaned):
        valid_numbers.append(cleaned)

print(valid_numbers)
```
