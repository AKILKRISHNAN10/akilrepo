## 2) Check palindrome string

```python
x=input('Enter the string :')
x1=list(x)
palindrome = x[::-1]

if x == palindrome:
    print(x ,' is a palindrome')
else:
    print(x ,' is not a palindrome')
```

## 3) Count vowels in string

```python
import  re

x=input('Enter the string')
x1 = re.findall('[aeiou]',x)
print(len(x1))
```

## 6) Find factorial

```python
x = int(input('enter the number : '))

factorial = 1
for i in range(2,x+1):
    factorial *= i
            
print('Factorial of the given number :',factorial)

    
```

## 7) Count even and odd numbers

```python
d=[1,2,3,4,5,6,7,8,10,-12,-965,-85,36.23,78/5]
even=[]
odd=[]
for i in d:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print('even numbers = ',even,'\nNumber of even numbers : ',len(even),'\n')
print('odd numbers = ',odd,'\nNumber of odd numbers : ',len(odd),'\n')
```
