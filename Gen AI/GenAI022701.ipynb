```python
print("\n================= 12. Quantifiers =================")

import re
#How many times a character or pattern should repeat

#| Symbol | Meaning            |
#| ------ | ------------------ |
#| `+`    | One or more times  |I need at least one.
#| `*`    | Zero or more times |Even zero is okay.
#| `?`    | Zero or one time   |Maybe yes, maybe no.

print(re.findall("a+", "caaaandy")) 
print(re.findall("aal*", "goaaaal")) #(pattern, text) 
print(re.findall("ba?", "bat ball"))

result = re.findall("l*", "goooal")
print(len(result))
print(result)
```

```python
import re

print("\n================= 1. re.match() =================")

#re.match() checks ONLY from position 0 (start of string)

text = "Python is powerful"
match_result = re.match('Python', text)
print("Match:", match_result)
print("Group:", match_result.group() if match_result else None)
```

# ======= re.sub / re.subn =======

```python
text63 = 'cat, mat, bat, hat'

for i in re.finditer('at',text63):
    print('iter',i.group(),'span',i.span())

print('\n ========== re.sub ========\n')

text64 = 'cat, mat, bat, hat'
t641 = re.sub('a','o',text64)
print(t641)

print('\n ========== re.subn ========\n')

text64 = 'cat, mat, bat, hat'
t641 = re.subn('a','o',text64)
print(t641)
```
